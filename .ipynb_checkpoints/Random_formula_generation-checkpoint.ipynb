# Random LTL formula generation

Generates `F` random formulas for every `(N, L)` combination (`N` propositions,
`L` symbols) and writes them to `FORMULAS_FILE`, one per line as
`N<TAB>L<TAB>index<TAB>formula`. Run the experiments on them with
`Random_formula_experiment.ipynb`.

```python
import random

alphabet = ['a','b','c','d','e','f','g','h','i','j','k','l','m','n','o','p','q','r','s','t','u','v','w','x','y','z']
unary = ['!', 'X ']
binary_non_temporal = ['&&', '||']
binary_temporal = ['U','W']
```

```python
P = 1/4
weights = [(1-P)/4]*4 + [P/2]*2
```

```python
# Sweep parameters
N_VALUES = [1, 2, 3]
L_VALUES = [5, 10, 15, 20]
F = 100  # number of random formulas per (N, L) combination

# None = a different set on every run; an int makes the set reproducible
SEED = None

FORMULAS_FILE = "formulas.txt"

# generate_formula() reads N from the global scope; it is reset inside the sweep
N = N_VALUES[0]
```

```python
def generate_formula(L):
    if L == 1:
        return random.choice(alphabet[:N])

    op = random.choices(unary+binary_temporal+binary_temporal, weights)[0]
    if L == 2:
        op = random.choice(unary)

    if op in unary:
        return '('+op+generate_formula(L-1)+')'

    S = random.randrange(1,L-1)

    return '(' + generate_formula(S) + ' ' + op + ' ' + generate_formula(L-S-1) + ')'
```

```python
random.seed(SEED)

formulas = []  # (N, L, index, formula)
for n in N_VALUES:
    N = n  # used by generate_formula
    for L in L_VALUES:
        formulas.extend((n, L, i, generate_formula(L)) for i in range(F))

with open(FORMULAS_FILE, "w") as fh:
    for n, L, i, formula in formulas:
        fh.write(f"{n}\t{L}\t{i}\t{formula}\n")

print(f"Wrote {len(formulas)} formulas to {FORMULAS_FILE}")
```
